```python
import time

def inefficient_sum(n):
    start_time = time.time()
    total = 0
    for i in range(1, n + 1):
        total += i
    end_time = time.time()
    print(f"Sum: {total}")
    print(f"Execution Time: {end_time - start_time:.6f} seconds")

n = int(input("Enter a number: "))
inefficient_sum(n)
```

```python
import time

def optimized_sum(n):
    start_time = time.time()
    total = n * (n + 1) // 2  # Formula for the sum of the first n numbers
    end_time = time.time()
    print(f"Sum: {total}")
    print(f"Execution Time: {end_time - start_time:.6f} seconds")

n = int(input("Enter a number: "))
optimized_sum(n)
```
