# ARRAY WITH RANDOM NUMBERS

## rand() for 1D: generate a random value between 0 to 1

```python
import numpy as np

arr_rand = np.random.rand(4)
print(arr_rand)
```

## rand() for 2D:

```python
arr_rand2 = np.random.rand(2,5)
print(arr_rand2)
```

## rand() for 3D:

```python
arr_rand3 = np.random.rand(1,2,3)
print(arr_rand3)
```

## randn(): generate a random value close to 0, either positive or negative

```python
arr_randn = np.random.randn(3)
print(arr_randn)
```

## ranf():  the function for doing random sampling in numpy. It returns an array of specified shape and fills it with random floats in the half open interval [0.0, 1.0).= '[' it means involved i.e. 0.0 is involved BUT ')' it means less than 1.0 

```python
arr_ranf = np.random.ranf(3)
print(arr_ranf)
```

## randint(): the function is used to generate a random number between a given range

```python
arr_4 = np.random.randint(4,24,6)
print(arr_4)
```
