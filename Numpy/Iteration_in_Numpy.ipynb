# 1D

```python
import numpy as np
a1 = np.array([1,2,3,4,5,6])

for i in a1:
    print(i)
```

# 2D

```python
a2 = np.array([[1,2,3,4], [5,6,7,8]])

for i in a2:
    for j in i:
        print(j)
```

# 3D

```python
a3 = np.array([[[1,2,3],[4,4,3],[4,1,3]]])
for i in a3:
    for j in i:
        for k in j:
            print(k)
```

# Iteration using function

```python
a4 = np.array([[[1,2,3],[9,3,6],[8,4,2]]])
for i in np.nditer(a4):
    print(i)
```

#### If we want to print the iterations with indexing number then  we have another function called np.ndenumerate()

```python
a5 = np.array([1,2,3,4,5])
for i,d in np.ndenumerate(a5):
    print(i,d)
```

```python
a6 = np.array([[[1,2,3],[8,45,6],[4,6,9]]])
for i,d in np.ndenumerate(a6):
    print(i,d)
```
