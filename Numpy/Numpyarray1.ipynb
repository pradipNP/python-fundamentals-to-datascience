Array

```python
import numpy as np
x = [2,3,4,5]
print(x)
print(type(x))
```


Convert list into array by taking input from user

```python
list = []
for i in range(1,6):
    num = int(input("Enter number: "))
    list.append(num)

arr = np.array(list)
print(arr)
print(type(arr))
print(arr.ndim) #To check the dimension either 1D, or 2D, or 3D, or Multidimensional
```


2D ARRAY

```python
a = np.array([[1,2,3], [4,5,6]])
print(a)
print(type(a))
print(a.ndim)
```


3D ARRAY

```python
b = np.array([[[1,2,3], [4,5,6], [2,4,2]]])
print(b)
print(type(b))
print(b.ndim)
```


N DIMENSION

```python
n = np.array([2,2,3,4], ndmin=10) #ndmin is used for dimension declaration
print(n)
print(type(n))
print(n.ndim) #ndim is used for dimension checking
```


ALL ELEMENTS OF ARRAY = 0

```python
import numpy as np
arr_zero1d = np.zeros(3)
arr_zero2d = np.zeros((3,4))
arr_zero3d = np.zeros(((3,4,5)))
print(arr_zero1d)
print()
print(arr_zero2d)
print()
print(arr_zero3d)
```


ALL ELEMENTS OF ARRAY = 1

```python
arr_one1d = np.ones(3)
arr_one2d = np.ones((3,4))
arr_one3d = np.ones(((3,4,5)))
print(arr_one1d)
print()
print(arr_one2d)
print()
print(arr_one3d)
print()
```


EMPTY ARRAY CREATE

```python
arr_empty = np.empty(4)
print(arr_empty)
```



ARANGE FUNCTION- To create an array in any range

```python
arr_arange = np.arange(4)
print(arr_arange)
```



DIAGONAL WITH 1 ARRAY CREATION

```python
arr_diagonal = np.eye(3)
print(arr_diagonal)
```



LINESPACE- Creation of array with some gap

```python
arr_space = np.linspace(0,20,num=5)
print(arr_space)
```
