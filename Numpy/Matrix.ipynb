# Matrix

```python
import numpy as np
m = np.matrix([[1,2],[4,3]])
n = np.matrix([[1,2],[4,5]])
print(m*n)
print()
print(m.dot(n))
```

# Array

```python
a = np.array([[1,2],[4,3]])
b = np.array([[1,2],[4,5]])
print(a*b)
```

# Matrix Function
#### Transpose
#### Swapaxes
#### Inverse
#### Power
#### Determinate

### Transpose

```python
m1 = np.matrix([[1,2,3],[4,5,6]])
print(m1)
print()
print(np.transpose(m1))
print()
# Or we can use another function for the transpose
print(m1.T)
print()
```

# Swapaxes

```python
#swapaxes() also swaps the axis
print()
print(np.swapaxes(m1,0,1)) # covert axis 0 to 1
```

# Inverse

```python
z= np.matrix([[1,2],[3,4]])
print(np.linalg.inv(z))
```

# Power

```python
z1 = np.matrix([[1,2],[3,5]])
print(z1)
print()

print(np.linalg.matrix_power(z1,2))

# To print indentity matrix,
print()
print(np.linalg.matrix_power(z1,0))
```

# Determinate

```python
z1 = np.matrix([[1,2],[3,5]])
print(z1)
print()
print(np.linalg.det(z1))
```
