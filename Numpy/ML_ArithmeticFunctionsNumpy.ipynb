# Shuffle

```python
import numpy as np
arr = np.array([1,2,3,4,5,6])
np.random.shuffle(arr)
print(arr)
```

# Unique

```python
arr2 = np.array([1,2,2,3,4,5,1,2,5])
u = np.unique(arr2) # unique function removes all duplicates
print(u)
print()

#if we want index numbers too then
u1 = np.unique(arr2, return_index=True, return_counts=True)
print(u1)
```

# Resize

```python
arr3 = np.array([1,2,3,4,5,6])
print(np.resize(arr3,(3,2)))
```

# Flatten & Ravel

```python
var = np.array([[1,2,3],[5,3,4]])
print("Flatten : ", var.flatten(order="F")) #converts 2d array in single
print("Ravel: ", np.ravel(var,order="A"))
```
