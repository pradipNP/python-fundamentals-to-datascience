# Datatype

```python
import numpy as np
arr = np.array([1,2,3])
print(arr)
print(arr.dtype)
```

### Datatype conversion

```python
arr1 = np.array([1,2,3], dtype = np.int8)
print(arr1)
print(arr1.dtype)
```

### Shortcut datatype conversion

```python
arr1 = np.array([1,2,3], dtype = "f")
print(arr1)
print(arr1.dtype)
```
