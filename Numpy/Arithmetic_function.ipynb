# ARITHMETIC FUNCTION

##### np.min(x)
##### np.max(x)
##### np.argmin(x) = shows the position of min
##### np.argmax(x) = shows the position of max
##### np.sqrt(x)
##### np.sin(x)
##### np.cos(x)
##### np.cumsum(x) = returns 1st element same then cross wise summation 

```python
import numpy as np
arr = np.array([1,2,3,4,5,3,2,5,2,1])
print("Minimum: ",np.min(arr))
print("Maximum: ",np.max(arr))
print("Position of min: ",np.argmin(arr))
print("Position of max: ",np.argmax(arr))
```

### 2D array

```python
ar = np.array([[1,2,3],[3,6,1]])
print("Min: ",np.min(ar,axis=0)) #axis=0 means column wise & axis=1 means row wise
print("Max: ",np.max(ar,axis=0))
print("Position of min: ",np.argmin(ar,axis=1))
print("Cumulative sum: ",np.cumsum(ar))
```

```python
arr1 = np.array([1,2,3])
print("Square root: ",np.sqrt(arr1))
```

```python
arr2 = np.array([0,30,60,90])
print("Sin values: ",np.sin(arr2))
print("Cos values: ",np.cos(arr2))
print("Cumulative sum: ", np.cumsum(arr2))
```
