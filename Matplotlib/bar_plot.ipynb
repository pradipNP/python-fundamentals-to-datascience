```python
import matplotlib.pyplot as plt
x = ["Python","Java","C", "C++", "SQL"]
y = [90, 70, 80, 65, 100]
plt.xlabel("Programming Language", fontsize =15)
plt.ylabel("Rate", fontsize = 15)
plt.title("Rating of Languages", fontsize = 20)
plt.bar(x,y)
plt.show()
```

# parameters in bar

```python

plt.bar(x,y, width = 0.4, color = "y")
```

```python
# 1.colors of each bar
# 2.align(i.e. position of bar)
c = ["r","y","b","m","k"]
plt.bar(x,y,width=0.5, color=c, align = "edge")
```

```python
plt.bar(x,y,width=0.5, color=c, align = "center")
```

```python
# edge color of bar
plt.bar(x,y,width=0.5, color=c, edgecolor = "k")
```

```python
#edge width
plt.bar(x,y,width=0.5, color=c, edgecolor = "k", linewidth = 5.4)
```

```python
# dotted linestyle
plt.bar(x,y,width=0.5, color=c, align = "edge", edgecolor = 'k', linewidth = 5, linestyle = ':')
```

```python
# hardness or softness of color of bar
plt.bar(x,y,width=0.5, color=c, align = "edge", edgecolor = 'k', linewidth = 5, alpha=0.2)
```

## Label using legend

```python
plt.xlabel("Programming Language", fontsize =15)
plt.ylabel("Rate", fontsize = 15)
plt.title("Rating of Languages", fontsize = 20)
plt.bar(x,y,width=0.5, color=c, align = "edge", edgecolor = 'k', label = "Programs")
plt.legend()
plt.show()
```

# Double bar 

```python
z = [10, 20, 30, 40, 50]
plt.xlabel("Programming Language", fontsize =15)
plt.ylabel("Rate", fontsize = 15)
plt.title("Rating of Languages", fontsize = 20)
plt.bar(x,y,width=0.5, color='b',label = "Programs")
plt.bar(x,z,width=0.5, color='r',label = "Subjects")
plt.legend()
plt.show()
```
