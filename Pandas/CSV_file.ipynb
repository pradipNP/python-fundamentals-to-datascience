# How to create CSV file

```python
import pandas as pd
dic = {"a":[1,2,3,4], "b":[1,2,3,4]}

var = pd.DataFrame(dic)
var["sum"] = var["a"]+var["b"]
var

var.to_csv("sum_of_a&b.csv")
var.to_csv("sum_without_index.csv", index=False)
var.to_csv("header_change.csv", index = False, header = ["A", "B", "C"])
var
```
