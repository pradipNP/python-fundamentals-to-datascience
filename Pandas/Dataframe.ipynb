# Dataframe

```python
import pandas as pd
l = [1,2,3,4,5,6]
var = pd.DataFrame(l)
print(type(var))
print()
var
```

## Dictionary

```python
dec = {"age":[20,22,23,26,19],"marks":[85,45,98,44,67]}
var1 = pd.DataFrame(dec)
var1

```

```python
dec2 = {"age":[20,22,23,26],"marks":[85,45,98,44], 1:[1,2,3,4], "a":['a','s','d','f']}
var2 = pd.DataFrame(dec2, columns =["a",1,"age"], index=[1,2,3,4])
print()
var2
```

## How to get data?

```python
dec3 = {"age":[20,22,23,26],"marks":[85,45,98,44], 1:[1,2,3,4], "a":['a','s','d','f']}
var3 = pd.DataFrame(dec2)
print(var3)
print()

print(var3["marks"][2])
```

## Using list

```python
list = ([[1,2,3,4],[9,8,7,6]])
l = pd.DataFrame(list)
l
```

## Using series

```python
serv = {"s":pd.Series([1,2,3,4]), "p":pd.Series([9,3,45,5])}
d = pd.DataFrame(serv)
d
```
