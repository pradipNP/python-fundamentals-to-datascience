# Series

```python
import pandas as pd
a = [1,2,3,4,5]
s = pd.Series(a)
print(s)
```

```python
b = [1,2,3,4,5]
s1 = pd.Series(b, index=['a','b','c','d','e'], dtype="float",name="Data")
s1
```

## Dictionary example

```python
dec = {
    "name" : ["Python", "C", "c++", "java"],  
    "por":[12,13,14,15],
    "rank":[1,4,3,2]
}
v = pd.Series(dec)
v
```

```python
s = pd.Series(12,index=[1,2,3,4])
s
```

## Addition of two diff. dimension's data which is not possible in NumPy

```python
s1 = pd.Series(12, index=[1,2,3,4,5,6,7])
s2 = pd.Series(13, index=[1,2,3,4])
print(s1+s2)
```
